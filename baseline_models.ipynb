```python
def data_preprocessing(full_df, test_size=0.2, random_state=42):
    train_df, test_df = train_test_split(
        full_df, test_size=test_size, random_state=random_state, stratify=full_df['label']
        )

    X_train = train_df.iloc[:, 1:]
    y_train = train_df['label']

    X_test = test_df.iloc[:, 1:]
    y_test = test_df['label']

    # already scaled
    # scaler = StandardScaler()
    # X_train = scaler.fit_transform(X_train)
    # X_test = scaler.transform(X_test)

    return X_train, X_test, y_train, y_test
```

```python
# Tuning

def tune_logreg(full_df, test_size=0.2, random_state=42, n_iter=10, n_splits=5):
    X_train, X_test, y_train, y_test = data_preprocessing(
        full_df, test_size=test_size, random_state=random_state
    )

    param_dist = {
        'C': np.linspace(0.5, 10, 5),
        'solver': ['saga'],
        'class_weight': [None, 'balanced'],
    }

    model = LogisticRegression(max_iter=1500)
    skf = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=random_state)


    random_search = RandomizedSearchCV(
        estimator=model,
        param_distributions=param_dist,
        n_iter=n_iter,
        cv=skf,
        verbose=0,
        n_jobs=-1,
        random_state=random_state,
    )
    random_search.fit(X_train, y_train)

    return random_search.best_estimator_, random_search.best_params_, X_test, y_test
```

```python
# Build + score one model

def build_logreg_model(full_df, test_size=0.2, random_state=42):

    best_model, best_params, X_test, y_test = tune_logreg(
        full_df,
        test_size=test_size,
        random_state=random_state
    )

    y_pred = best_model.predict(X_test)
    y_prob = best_model.predict_proba(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test, y_pred, zero_division=0, average='weighted'
    )

    recall = recall_score(
        y_test, y_pred, zero_division=0, average='weighted'
    )

    f1 = f1_score(
        y_test, y_pred, zero_division=0, average='weighted'
    )

    auc = roc_auc_score(
        y_test,
        y_prob,
        multi_class='ovr',
        average='weighted'
    )

    # Multiclass coefficient importance
    coefficients = best_model.coef_

    importance = pd.DataFrame({
        'Feature': X_test.columns,
        'Mean_Abs_Coefficient': np.mean(
            np.abs(coefficients), axis=0
        )
    }).sort_values(
        'Mean_Abs_Coefficient',
        ascending=False
    ).reset_index(drop=True)

    return best_model, best_params, accuracy, precision, recall, f1, auc, importance
```

```python
# Replicate across different participant splits

def logreg_wrap(full_df, n_replicates=30, test_size=0.2):
    rows = []
    importance_frames = []

    for i in range(n_replicates):
        print(f'Replicate {i}')
        (best_model, best_params, accuracy, precision,
         recall, f1, auc, importance) = build_logreg_model(
            full_df, test_size=test_size, random_state=i
        )

        row = {
            'Accuracy': accuracy,
            'Precision': precision,
            'Recall': recall,
            'F1-score': f1,
            'AUC': auc,
        }
        row.update(best_params)
        rows.append(row)

        importance_frames.append(importance.assign(Replicate=i))

    results_df = pd.DataFrame(rows)


    # Aggregate importance across replicates
    all_importance = pd.concat(importance_frames, ignore_index=True)
    importance_df = (
    all_importance
    .groupby('Feature')
    .agg(
        Mean_Abs_Coefficient=('Mean_Abs_Coefficient', 'mean'),
        Std_Abs_Coefficient=('Mean_Abs_Coefficient', 'std')
    )
    .sort_values('Mean_Abs_Coefficient', ascending=False)
    .reset_index()
)

    return results_df, importance_df
```

```python
def data_preprocessing(full_df, test_size=0.2, random_state=42):
    train_df, test_df = train_test_split(
        full_df,
        test_size=test_size,
        random_state=random_state,
        stratify=full_df['label']
    )

    X_train = train_df.iloc[:, 1:]
    y_train = train_df['label']

    X_test = test_df.iloc[:, 1:]
    y_test = test_df['label']

    return X_train, X_test, y_train, y_test
```

```python

def tune_rfc(full_df, test_size=0.2, random_state=42,
             n_iter=15, n_splits=5):

    X_train, X_test, y_train, y_test = data_preprocessing(
        full_df, test_size=test_size, random_state=random_state
    )

    param_dist = {
        'n_estimators': [100, 200, 300, 500],
        'max_depth': [None, 10, 20, 30, 50],
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4],
        'max_features': ['sqrt', 'log2'],
        'class_weight': [None, 'balanced', 'balanced_subsample']
    }

    model = RandomForestClassifier(
        random_state=random_state,
        n_jobs=-1
    )

    skf = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=random_state
    )

    random_search = RandomizedSearchCV(
        estimator=model,
        param_distributions=param_dist,
        n_iter=n_iter,
        cv=skf,
        scoring='f1_macro',
        verbose=0,
        n_jobs=-1,
        random_state=random_state
    )

    random_search.fit(X_train, y_train)

    return (
        random_search.best_estimator_,
        random_search.best_params_,
        X_test,
        y_test
    )
```

```python

def build_rfc_model(full_df, test_size=0.2, random_state=42):

    best_model, best_params, X_test, y_test = tune_rfc(
        full_df,
        test_size=test_size,
        random_state=random_state
    )

    y_pred = best_model.predict(X_test)
    y_prob = best_model.predict_proba(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test, y_pred, zero_division=0, average='weighted'
    )

    recall = recall_score(
        y_test, y_pred, zero_division=0, average='weighted'
    )

    f1 = f1_score(
        y_test, y_pred, zero_division=0, average='weighted'
    )

    auc = roc_auc_score(
        y_test,
        y_prob,
        multi_class='ovr',
        average='weighted',
        labels=best_model.classes_
    )

    # Feature importance
    importance = pd.DataFrame({
        'Feature': X_test.columns,
        'Importance': best_model.feature_importances_
    }).sort_values(
        'Importance',
        ascending=False
    ).reset_index(drop=True)

    return (
        best_model, best_params, accuracy,
        precision, recall, f1, auc, importance
    )
```

```python

def rfc_wrap(full_df, n_replicates=30, test_size=0.2):

    rows = []
    importance_frames = []

    for i in range(n_replicates):

        print(f'Replicate {i}')

        (best_model, best_params, accuracy, precision,
         recall, f1, auc, importance) = build_rfc_model(
            full_df,
            test_size=test_size,
            random_state=i
        )

        row = {
            'Accuracy': accuracy,
            'Precision': precision,
            'Recall': recall,
            'F1-score': f1,
            'AUC': auc
        }

        row.update(best_params)
        rows.append(row)

        importance_frames.append(
            importance.assign(Replicate=i)
        )

    results_df = pd.DataFrame(rows)

    # Aggregate feature importance
    all_importance = pd.concat(
        importance_frames, ignore_index=True
    )

    importance_df = (
        all_importance
        .groupby('Feature')
        .agg(
            Mean_Importance=('Importance', 'mean'),
            Std_Importance=('Importance', 'std')
        )
        .sort_values(
            'Mean_Importance', ascending=False
        )
        .reset_index()
    )

    return results_df, importance_df
```

```python
def tune_xgb(full_df, test_size=0.2, random_state=42,
             n_iter=15, n_splits=5):

    X_train, X_test, y_train, y_test = data_preprocessing(
        full_df, test_size=test_size, random_state=random_state
    )

    # Encode target labels for XGBoost
    le = LabelEncoder()
    y_train = le.fit_transform(y_train)
    y_test = le.transform(y_test)

    param_dist = {
        'n_estimators': [100, 200, 300, 500],
        'max_depth': [3, 4, 5, 6, 8],
        'learning_rate': [0.01, 0.03, 0.05, 0.1, 0.2],
        'subsample': [0.7, 0.8, 0.9, 1.0],
        'colsample_bytree': [0.7, 0.8, 0.9, 1.0],
        'min_child_weight': [1, 3, 5, 7],
        'gamma': [0, 0.1, 0.3, 0.5],
        'reg_lambda': [1, 2, 5, 10]
    }

    model = XGBClassifier(
        objective='multi:softprob',
        num_class=len(le.classes_),
        eval_metric='mlogloss',
        random_state=random_state,
        n_jobs=1
    )

    skf = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=random_state
    )

    random_search = RandomizedSearchCV(
        estimator=model,
        param_distributions=param_dist,
        n_iter=n_iter,
        cv=skf,
        scoring='f1_macro',
        verbose=0,
        n_jobs=-1,
        random_state=random_state
    )

    random_search.fit(X_train, y_train)

    return (
        random_search.best_estimator_,
        random_search.best_params_,
        X_test,
        y_test
    )
```

```python

def build_xgb_model(full_df, test_size=0.2, random_state=42):

    best_model, best_params, X_test, y_test = tune_xgb(
        full_df,
        test_size=test_size,
        random_state=random_state
    )

    y_pred = best_model.predict(X_test)
    y_prob = best_model.predict_proba(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test, y_pred, zero_division=0, average='weighted'
    )

    recall = recall_score(
        y_test, y_pred, zero_division=0, average='weighted'
    )

    f1 = f1_score(
        y_test, y_pred, zero_division=0, average='weighted'
    )

    auc = roc_auc_score(
        y_test,
        y_prob,
        multi_class='ovr',
        average='weighted'
    )

    # Feature importance
    importance = pd.DataFrame({
        'Feature': X_test.columns,
        'Importance': best_model.feature_importances_
    }).sort_values(
        'Importance',
        ascending=False
    ).reset_index(drop=True)

    return (
        best_model, best_params, accuracy,
        precision, recall, f1, auc, importance
    )
```

```python

def xgb_wrap(full_df, n_replicates=30, test_size=0.2):

    rows = []
    importance_frames = []

    for i in range(n_replicates):

        print(f'Replicate {i}')

        (best_model, best_params, accuracy, precision,
         recall, f1, auc, importance) = build_xgb_model(
            full_df,
            test_size=test_size,
            random_state=i
        )

        row = {
            'Accuracy': accuracy,
            'Precision': precision,
            'Recall': recall,
            'F1-score': f1,
            'AUC': auc
        }

        row.update(best_params)
        rows.append(row)

        importance_frames.append(
            importance.assign(Replicate=i)
        )

    results_df = pd.DataFrame(rows)

    # Aggregate feature importance
    all_importance = pd.concat(
        importance_frames,
        ignore_index=True
    )

    importance_df = (
        all_importance
        .groupby('Feature')
        .agg(
            Mean_Importance=('Importance', 'mean'),
            Std_Importance=('Importance', 'std')
        )
        .sort_values(
            'Mean_Importance', ascending=False
        )
        .reset_index()
    )

    return results_df, importance_df
```

### Final Model

```python

import joblib
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import (
    RandomizedSearchCV,
    StratifiedKFold
)


def train_final_rfc(
    full_df,
    id_columns=None,
    random_state=42,
    n_iter=15,
    n_splits=5,
    save_path="final_baseline_model/asl_random_forest.joblib"
):

    id_columns = id_columns or []

    # Separate features and target
    feature_columns = [
        col for col in full_df.columns
        if col not in ['label'] + id_columns
    ]

    X = full_df[feature_columns]
    y = full_df['label']

    # Hyperparameter search
    param_dist = {
        'n_estimators': [100, 200, 300, 500],
        'max_depth': [None, 10, 20, 30, 50],
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4],
        'max_features': ['sqrt', 'log2'],
        'class_weight': [
            None, 'balanced', 'balanced_subsample'
        ]
    }

    model = RandomForestClassifier(
        random_state=random_state,
        n_jobs=-1
    )

    skf = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=random_state
    )

    search = RandomizedSearchCV(
        estimator=model,
        param_distributions=param_dist,
        n_iter=n_iter,
        cv=skf,
        scoring='f1_macro',
        verbose=1,
        n_jobs=-1,
        random_state=random_state,
        refit=True
    )

    # Train using all available labeled data
    search.fit(X, y)

    final_model = search.best_estimator_

    # Save model and metadata together
    model_bundle = {
        'model': final_model,
        'feature_columns': feature_columns,
        'classes': final_model.classes_,
        'best_params': search.best_params_
    }

    joblib.dump(model_bundle, save_path)

    print(f"Model saved to: {save_path}")
    print(f"Best parameters: {search.best_params_}")

    return final_model, search.best_params_
```
